Membuat Program Python untuk Meminta pengguna memasukkan nama dan umur.

```python
nama = input("Masukan Nama :")
umur = int(input("Masukan Umur :"))

print ("Nama :" , nama)
print ("Umur :" , umur)
```

Baris 4 : Fungsi untuk mencetak teks “Nama :”, lalu diikuti nilai dari variabel nama

Membuat Program untuk Menghitung tahun lahir pengguna (dari umur).

```python
umur_string = input("Masukkan Umur Anda : ")
umur_int = int(umur_string)
tahun_sekarang = 2026
tahun_lahir = tahun_sekarang - umur_int
print(f"Jika umur anda {umur_int} tahun, tahun kelahiran anda {tahun_lahir}")
```

Baris 1 : Membuat fungsi input untuk meminta pengguna menginput umur di terminal, variabel yang disimpan adalah berbentuk string

Baris 2 : Mengonversi Teks dari umur string, menjadi umur integer dan disimpan di umur_int

Baris 3 : Menentukan nilai variabel yaitu tahun saat ini

Baris 4 : Membuat Rumus untuk menghitung tahun lahir, didapat dari tahun sekarang dikurangi dengan umur_int/input pengguna

Baris 5 : Menampilkan (Print) hasil perhitungan menggunakan f-string, dan menyisipkan variabel ditengahnya
