```python
'''
Membuat program CRUD – Koleksi Lagu
• Tambahkan lagu baru (judul, penyanyi).
• Tampilkan semua lagu.
• Hapus lagu berdasarkan judul.
• Cari lagu berdasarkan nama penyanyi.
'''

koleksi_lagu = []

#Fungsi Menambahkan lagu ke koleksi (judul dan penyanyi)
def tambah_lagu(judul, penyanyi):
    koleksi_lagu.append({"judul": judul, "penyanyi": penyanyi})
    print(f"'{judul}' oleh {penyanyi} masuk ke koleksi")

#Fungsi Menampilkan lagu Koleksi
def tampilkan_lagu():
    if not koleksi_lagu:
        print("Belum ada di koleksi :(")
    else:
        print("\n---- Daftar Lagu Arif ----")
        for i, lagu in enumerate(koleksi_lagu, start=1):
            print(f"{i}. {lagu['judul']} - {lagu['penyanyi']}")

#Fungsi Menghapus Lagu berdasarkan judul
def hapus_lagu(judul):
    global koleksi_lagu
    lagu_ditemukan = [lagu for lagu in koleksi_lagu if lagu["judul"].lower() == judul.lower()]

    if not lagu_ditemukan:
        print(f"Waduh, lagu'{judul}' tidak ada dikoleksi")
        return

    koleksi_lagu = list(filter(lambda lagu: lagu["judul"].lower() != judul.lower(), koleksi_lagu))
    print(f"Lagu '{judul}' sudah dihapus")

#Fungsi Untuk Mencari Lagu berdasarkan penyanyi
def cari_lagu(penyanyi):
    hasil = list(filter(lambda lagu: penyanyi.lower() in lagu["penyanyi"].lower(), koleksi_lagu))
    return hasil

#Membuat tampilan menu utama dan untuk input ke terminal
def menu():

    while True:
        print("\n--- KOLEKSI LAGU ---")
        print("1.Mau Menambah Lagu Ke Koleksi")
        print("2.Menampilkan lagu di Koleksi")
        print("3.Mengapus Lagu")
        print("4.Mencari Lagu berdasarkan Penyanyi")
        print("5.Keluar")

        pilihan = input("Pilih antara (1-5): ")

        if pilihan == "1":
            judul = input("Judul lagu yang ditambah: ")
            penyanyi = input("Nama Penyanyi: ")
            tambah_lagu(judul, penyanyi)

        elif pilihan == "2":
            tampilkan_lagu()

        elif pilihan == "3":
            judul = input("Lagu yang ingin dihapus: ")
            hapus_lagu(judul)

        elif pilihan == "4":
            penyanyi = input("Penyanyi yang dicari")
            hasil = cari_lagu(penyanyi)
            if hasil:
                print(f"\nSudah menemukan {len(hasil)} lagu di koleksi:")
                for lagu in hasil:
                    print(f"- {lagu['judul']} oleh {lagu['penyanyi']}")
            else:
                print(f"Belum ada lagu dengan penyanyi '{penyanyi}'.")

        elif pilihan == "5":
            print("Program Selesai")
            break

        else:
            print("Pilihan tidak valid,ulangi lagi")


if __name__ == "__main__":
    tambah_lagu("Multo", "Cup of Joe")
    tambah_lagu("About You", "The 1975")
    tambah_lagu("Shape of My Heart", "Backstreet Boys")

    menu()
```
