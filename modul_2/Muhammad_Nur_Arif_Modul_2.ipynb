```python
#Soal No 1 : Membuat Class Dosen
class Dosen:
    def __init__(self, nama, nidn, mata_kuliah):
        self.nama = nama
        self.nidn = nidn
        self.mata_kuliah = mata_kuliah

#Soal no 2 Method Info
    def info(self):
        print(f"Nama: {self.nama}, NIDN: {self.nidn}, Mata Kuliah: {self.mata_kuliah}")

#Soal no 3 Method Update Mata Kuliah
    def update_mata_kuliah(self, mk_baru):
        print(f"{self.nama} mengajar mata kuliah {mk_baru} (sebelumnya mengajar {self.mata_kuliah})")
        self.mata_kuliah = mk_baru


#Soal no 4 untuk membuat 3 object dosen, lalu  menampilkan datanya
dosen1 = Dosen("Prof Dr. Agus Efendi M.Pd", "0019086701", "Rangkaian Elektronika dan Intrumentasi")
dosen2 = Dosen("Dr. Aris Budianto S.T., M.Eng ", "0017128005", "Strategi Belajar Mengajar")
dosen3 = Dosen("Vearen Dika Sofirudin S.Pd., M.Ed", "5940777678230142", "Desain Web")
dosen1.info()
dosen2.info()
dosen3.info()
dosen1.update_mata_kuliah("Evaluasi Pendidikan")
dosen1.info()
```
